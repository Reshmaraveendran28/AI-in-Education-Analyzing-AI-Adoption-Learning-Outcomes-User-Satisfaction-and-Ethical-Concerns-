## Project Overview

Artificial Intelligence is increasingly becoming part of the modern learning environment. Students and learners use AI-powered tools for activities such as solving problems, receiving explanations, completing tasks, and accessing personalized learning support.

This project explores how AI tool usage relates to learning outcomes, engagement, user satisfaction, trust, sentiment, and ethical concerns. Using Python-based data analysis and visualization, the project aims to identify meaningful patterns and relationships within the dataset.

### Checking for Missing Values

After examining the basic structure and statistical summary of the dataset, the next step is to assess its data quality. First, the dataset will be checked for missing values in each column to determine whether any variables require further preprocessing.


```python
# Calculate the total number of missing values in each column
missing_values = df.isnull().sum()

# Display the missing-value count for each column
print("Missing values in each column:")
print(missing_values)
```

### Assessing Missing Value Percentage

The percentage of missing values will be calculated for each column to understand the extent of missing data relative to the total number of records.


```python
# Calculate the percentage of missing values in each column
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Create a summary table for missing-value analysis
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

# Display only columns containing missing values
missing_summary[missing_summary['Missing Values'] > 0].sort_values(
    by='Missing Percentage',
    ascending=False
)
```

### Missing Value Assessment — Result

The missing-value analysis shows that there are **no missing values in the dataset**. Therefore, all records contain complete information across the available variables, and no missing-value treatment or imputation is required at this stage.

In this step, the data types of the dataset columns are checked to ensure that numerical, categorical, and date-related values are stored in the appropriate format. The timestamp column is converted into datetime format for accurate time-based analysis.

```python
# Check the current data types
print("Data types before conversion:")
print(df.dtypes)

# Convert timestamp to datetime format
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')

# Verify the converted data type
print("\nData type of timestamp after conversion:")
print(df['timestamp'].dtype)
```

### Creating a Working Copy

A separate working copy of the dataset is created so that all cleaning and feature engineering operations can be performed without changing the original imported dataset.

**What was done:** A working copy named `df_clean` was created from the original dataset. All further cleaning and preprocessing operations will be performed on this working copy.

### Standardizing Categorical Values

Categorical values are standardized by removing unnecessary spaces and ensuring consistent text formatting. This helps avoid duplicate categories caused by differences in capitalization or extra spaces.

```python
# Identify categorical columns
categorical_columns = df_clean.select_dtypes(include='object').columns

print("Categorical columns:")
print(categorical_columns.tolist())

# Remove unnecessary spaces from categorical values
for column in categorical_columns:
    df_clean[column] = df_clean[column].astype(str).str.strip()

# Standardize sentiment labels to lowercase
df_clean['sentiment_label'] = df_clean['sentiment_label'].str.lower()

# Verify sentiment categories
print("\nStandardized sentiment labels:")
print(df_clean['sentiment_label'].unique())
```

**What was done:** The categorical values were standardized by removing unnecessary spaces and ensuring consistent text formatting. Sentiment labels were also converted to lowercase. The categories were then checked to confirm consistency.

### Checking for Unrealistic Values

In this step, important numerical variables are checked for values outside their expected realistic ranges. This helps identify invalid values that could negatively affect the analysis.

```python
# Check for unrealistic values in important numerical columns

print("Pre-test scores outside 0-100:",
      ((df_clean['pre_test_score'] < 0) |
       (df_clean['pre_test_score'] > 100)).sum())

print("Post-test scores outside 0-100:",
      ((df_clean['post_test_score'] < 0) |
       (df_clean['post_test_score'] > 100)).sum())

print("Sentiment polarity outside -1 to 1:",
      ((df_clean['sentiment_polarity'] < -1) |
       (df_clean['sentiment_polarity'] > 1)).sum())

print("Subjectivity outside 0-1:",
      ((df_clean['subjectivity'] < 0) |
       (df_clean['subjectivity'] > 1)).sum())
```

**What was done:** The important numerical variables were checked against their expected ranges. No invalid values were identified in the checked variables, so no corrections were required.

### Feature Cleaning – Extracting Hashtag Count

The `text_feedback` column is analyzed to identify hashtags in each feedback entry. A new `hashtag_count` feature is created to represent the number of hashtags present in each record.

```python
# Count hashtags in each feedback entry
df_clean['hashtag_count'] = df_clean['text_feedback'].str.count('#')

# Display the new feature
print(df_clean[['text_feedback', 'hashtag_count']].head())

# Check hashtag count distribution
print("\nHashtag count distribution:")
print(df_clean['hashtag_count'].value_counts().sort_index())
```

**What was done:** A new `hashtag_count` feature was created by counting hashtags in the `text_feedback` column. The dataset does not contain hashtags in the feedback text, so the hashtag count is zero. No artificial values were added.

### Cleaning Sentiment Labels

The sentiment labels are cleaned by removing unnecessary spaces and converting the labels to lowercase. The final unique values are checked to ensure that the sentiment categories are consistent.

```python
# Clean sentiment labels
df_clean['sentiment_label'] = (
    df_clean['sentiment_label']
    .astype(str)
    .str.strip()
    .str.lower()
)

# Check unique sentiment labels
print("Cleaned sentiment labels:")
print(df_clean['sentiment_label'].unique())

# Check sentiment distribution
print("\nSentiment distribution:")
print(df_clean['sentiment_label'].value_counts())
```
